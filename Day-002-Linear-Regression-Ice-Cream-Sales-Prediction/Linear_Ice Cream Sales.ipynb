```python
from sklearn.model_selection import train_test_split
```

```python
from sklearn.linear_model import LinearRegression
```

```python
x=df[['Temperature']]
y=df['Ice Cream Profits']
```

```python
xtrain,xtest,ytrain,ytest=train_test_split(x,y,test_size=0.2,random_state=0)
```

```python
model = LinearRegression()
model.fit(xtrain, ytrain)
```

```python
#scope=model.coef_
model.coef_
```

```python

#intercept=model.intercept_
model.intercept_
```

```python
model.predict(x)
```

```python
 #Plot Regression Line
plt.figure(figsize=(8,5))
plt.scatter(df["Temperature"], df["Ice Cream Profits"], color="blue", label="Actual Data")
plt.plot(df["Temperature"], model.predict(x), color="red", linewidth=2, label="Regression Line")
plt.title("Linear Regression - Ice Cream Sales Prediction")
plt.xlabel("Temperature")
plt.ylabel("Ice Cream Profits")
plt.legend()
plt.grid(True)
plt.show()
```
