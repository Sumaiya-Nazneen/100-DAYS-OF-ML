```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from xgboost import XGBClassifier
```

```python
# Load Dataset
df = pd.read_csv("Bank_Churn.csv")
```

```python
print(df.head())
```

```python
# Drop unnecessary columns
df = df.drop(columns=["RowNumber", "CustomerId", "Surname"])
```

```python
# Features and Target
x = df.drop(columns=["Exited"])
y = df["Exited"]
```

```python
# Categorical Columns
categorical_cols = ["Geography", "Gender"]
```

```python
# Preprocessing
preprocessor = ColumnTransformer(
    transformers=[
        (
            "onehot",
            OneHotEncoder(handle_unknown="ignore"),
            categorical_cols
        )
    ],
    remainder="passthrough"
)
```

```python
# Train-Test Split
xtrain,xtest,ytrain,ytest=train_test_split(x,y,train_size=0.8,random_state=42)
```

```python
# Pipeline
main_pipeline = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("model", XGBClassifier(
            n_estimators=100,
            max_depth=5,
            learning_rate=0.1,
            random_state=42
        ))
    ]
)
```

```python
# Train Model
main_pipeline.fit(xtrain, ytrain)
```

```python
# Prediction
y_pred = main_pipeline.predict(xtest)
```

```python
print("Accuracy:", accuracy_score(ytest, y_pred))
print(classification_report(ytest, y_pred))
print(confusion_matrix(ytest, y_pred))
```

# Conclusion

An XGBoost Classifier was successfully implemented on the Bank Customer Churn Prediction dataset to predict whether a customer will leave the bank. 

The unnecessary columns (RowNumber, CustomerId, and Surname) were removed. The categorical features (Geography and Gender) were converted into numerical values using OneHotEncoder with a ColumnTransformer, and a Pipeline was used to combine preprocessing and model training.

 The dataset was split into training and testing sets, predictions were made, and the model's performance was evaluated using Accuracy, Classification Report, and Confusion Matrix.

 . The model achieved an accuracy of 86.95%, demonstrating good performance in predicting whether a customer will stay or leave the bank
